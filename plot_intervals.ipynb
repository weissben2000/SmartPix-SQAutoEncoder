```python
import numpy as np
import pandas as pd

import matplotlib
import matplotlib.pyplot as plt

from plotting import *
```

```python
noise = [0,80, 120,240]

fig, ax = plt.subplots(1,3,figsize=(12,4),gridspec_kw={'wspace': 0.35,'hspace':0.3})

labels = ["SQ SlimMLP",
          "SQ SlimConv2D",
          "SQ Transformer",
          "SQAutoEncoded SlimMLP",
          "SQAutoEncoded SlimConv2D",
          "SQAutoEncoded Transformer",
         ]
training_names = ['SlimMLP_SQ',
                  'SlimConv2D_SQ',
                  'Transformer',
                  'SQAutoEncoder_SlimMLP',
                  'SQAutoEncoder_SlimConv2D',
                  'SQAutoEncoder_Transformer'
                 ]

for i,t in enumerate(labels):

    marker = 'o'
    # color=colors[i]
    ls = '-'
    
    if 'MLP' in t:        
        color = colors[1]
    elif 'Conv' in t:        
        color = colors[2]
    elif 'Trans' in t:        
        color = colors[0]

        
    # if 'SlimConv2D' in t:
    if 'SQ ' in t:        
        marker = 's'
        ls = ':'
        # color = 'black'

    x_mean = []
    y_mean = []
    B_mean = []
    A_mean = []

    x_std = []
    y_std = []
    B_std = []
    A_std = []
    
    for n in noise:
        name = training_names[i] + "_" + str(n) + "e_2bit"

        if "Transformer" in t and "SQ " in t:
            name = training_names[i] + str(n) + "e_2bit"
        filename = "avg_res_"+name+".parquet"
        # if not os.path(filename).e
        df = pd.read_parquet("avg_res_"+name+".parquet")

        x_mean += [np.mean(df['up68_x'] + df['down68_x'])]
        y_mean += [np.mean(df['up68_y'] + df['down68_y'])]
        B_mean += [np.mean(df['up68_B'] + df['down68_B'])]

        x_std += [np.std(df['up68_x'] + df['down68_x'])]
        y_std += [np.std(df['up68_y'] + df['down68_y'])]
        B_std += [np.std(df['up68_B'] + df['down68_B'])]

        # if "Slim" not in t:
        #     A_mean += [np.mean(df['up68_A'] + df['down68_A'])]
        #     A_std += [np.std(df['up68_A'] + df['down68_A'])]

    ax[0].set_ylabel("$I_x (68\%)$ [um]")
    ax[0].errorbar(x=noise, y=x_mean, yerr=x_std, linestyle=ls, marker=marker, ms=5,label=labels[i], color=color)

    ax[1].set_ylabel("$I_y (68\%)$ [um]")
    ax[1].errorbar(x=noise, y=y_mean, yerr=y_std, linestyle=ls, marker=marker, ms=5,label=labels[i], color=color)

    # if "Slim" not in t:
    #     ax[1][0].set_ylabel("$I_A(68\%)$ [deg]")
    #     ax[1][0].errorbar(x=noise, y=A_mean, yerr=A_std, linestyle=ls, marker=marker, ms=5,label=labels[i], color=color)

    
    a = ax[2].set_ylabel("$I_B (68\%)$ [deg]")
    ax[2].errorbar(x=noise, y=B_mean, yerr=B_std, linestyle=ls, marker=marker, ms=5,label=labels[i], color=color)

hands, legs = ax[0].get_legend_handles_labels()
fig.legend(hands, legs, loc='lower center', ncols=2, frameon=True, bbox_to_anchor=(0.5, -0.25))

ax[0].set_xlabel('Noise [e]')
ax[1].set_xlabel('Noise [e]')
ax[2].set_xlabel('Noise [e]')

ax[0].set_xticks(noise);
ax[1].set_xticks(noise);
ax[2].set_xticks(noise);
# ax[1][1].set_xticks(noise);

ax[0].grid(axis='y')
ax[1].grid(axis='y')
ax[2].grid(axis='y')
# ax[1][1].grid(axis='y')
fig.tight_layout()

plt.savefig('SQAutoEncoder_reg_performance_byNoise.png',bbox_inches='tight')
# plt.savefig('results3.pdf',bbox_inches='tight')
```
